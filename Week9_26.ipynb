# 3. Buatlah fungsi dilasi serta tampilkan hasil dilasi pada citra dengan Structuring Element Dilasi (SED) berbentuk cross 3 x 3 tanpa menggunakan library morphology dari openCV, sehingga menghasilkan tampilan seperti di bawah ini:

```python
# Dilasi tanpa library

def dilasi_citra(F,w):
  # size image
  p,q = F.shape

  imgD = np.zeros((p,q), dtype=np.uint8)

  # Generate structure element dilasi (SED)
  SED = np.ones((w,w), dtype=np.uint8)
  constant1 = (w-1)//2

  # Proses Dilasi
  for i in range(constant1, p-constant1):
    for j in range(constant1, q-constant1):
      temp = F[i-constant1:i+constant1+1, j-constant1:j+constant1+1]
      product = temp*SED
      imgD[i,j] = np.max(product)

  return imgD


img_noisy = cv2.imread('/content/drive/MyDrive/PCVK/Images/noisy_text.png',0)
ret, thresh = cv2.threshold(img_noisy, 125, 200, cv2.THRESH_BINARY)
w = 5
imgN = dilasi_citra(thresh,w)

img_plat = cv2.imread('/content/drive/MyDrive/PCVK/Images/plat_nomor.jpg',0)
ret, thresh_plat = cv2.threshold(img_plat, 125, 200, cv2.THRESH_BINARY)
w = 5
imgP = dilasi_citra(thresh_plat,w)

plt.subplot(2,2,1),plt.imshow(img_noisy,cmap='gray')
plt.title('Noisy'), plt.xticks([]), plt.yticks([])

plt.subplot(2,2,2),plt.imshow(imgN,cmap='gray')
plt.title('Dilasi (Noisy)'), plt.xticks([]), plt.yticks([])

plt.subplot(2,2,3),plt.imshow(img_plat,cmap='gray')
plt.title('Plat Nomor'), plt.xticks([]), plt.yticks([])

plt.subplot(2,2,4),plt.imshow(imgP,cmap='gray')
plt.title('Dilasi (Plat Nomor)'), plt.xticks([]), plt.yticks([])

plt.show()
```

# 5. Buatlah fungsi erosi serta tampilkan hasil erosi pada citra dengan Structuring Element Erosi (SE) berbentuk square 5 x 5 tanpa menggunakan library morphology dari openCV, sehingga menghasilkan tampilan seperti di bawah ini:

```python
#Erosi Tanpa Library

def erosi_citra(F,k):
  #size image
  m,n= F.shape

  #Generate structure element erosi (SE)
  SE= np.ones((k,k), dtype=np.uint8)
  constant= (k-1)//2

  imgE= np.zeros((m,n), dtype=np.uint8)

  #Proses Erosi
  for i in range(constant, m-constant):
    for j in range(constant,n-constant):
      temp= F[i-constant:i+constant+1, j-constant:j+constant+1]
      product= temp*SE
      imgE[i,j]= np.min(product)

  return imgE

img_noisy= cv2.imread('/content/drive/MyDrive/PCVK/Images/noisy_text.png',0) #
ret1, thresh_noisy = cv2.threshold(img_noisy, 175, 225, cv2.THRESH_BINARY)
k=5
imgN = erosi_citra(thresh_noisy,k)

img_plat= cv2.imread('/content/drive/MyDrive/PCVK/Images/plat_nomor.jpg',0) #
ret1, thresh_plat = cv2.threshold(img_plat, 175, 225, cv2.THRESH_BINARY)
k=5
imgP = erosi_citra(thresh_plat,k)

plt.subplot(2,2,1),plt.imshow(img_noisy,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])

plt.subplot(2,2,2),plt.imshow(imgN,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])

plt.subplot(2,2,3),plt.imshow(img_plat,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])

plt.subplot(2,2,4),plt.imshow(imgP,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])

plt.show()
```

```python
#Erosi Line Vertikal SE

def erosi_citra(F,k):
  #size image
  m,n= F.shape

  #Generate structure element erosi (SE)
  SE = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(1,k)) #SE line vertikal
  constant= (k-1)//2

  imgE= np.zeros((m,n), dtype=np.uint8)

  #Proses Erosi
  for i in range(constant, m-constant):
    for j in range(constant,n-constant):
      temp= F[i-constant:i+constant+1, j-constant:j+constant+1]
      product= temp*SE
      imgE[i,j]= np.min(product)

  return imgE
print('SE = Line Vertikal 3')
img_N= cv2.imread('/content/drive/MyDrive/PCVK/Images/noisy_text.png',0) #
ret1, thresh1 = cv2.threshold(img_N, 127, 225, cv2.THRESH_BINARY)
k=3
imgN = erosi_citra(thresh1,k)

img_P= cv2.imread('/content/drive/MyDrive/PCVK/Images/plat_nomor.jpg',0) #
ret1, thresh1 = cv2.threshold(img_P, 127, 225, cv2.THRESH_BINARY)
k=3
imgP = erosi_citra(thresh1,k)

plt.subplot(221),plt.imshow(img_N,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(222),plt.imshow(imgN,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])

plt.subplot(223),plt.imshow(img_P,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(224),plt.imshow(imgP,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])

plt.show()

print('SE = Line Vertikal 5')
img_N= cv2.imread('/content/drive/MyDrive/PCVK/Images/noisy_text.png',0) #
ret1, thresh1 = cv2.threshold(img_N, 127, 225, cv2.THRESH_BINARY)
k=5
imgN = erosi_citra(thresh1,k)

img_P= cv2.imread('/content/drive/MyDrive/PCVK/Images/plat_nomor.jpg',0) #
ret1, thresh1 = cv2.threshold(img_P, 127, 225, cv2.THRESH_BINARY)
k=5
imgP = erosi_citra(thresh1,k)

plt.subplot(221),plt.imshow(img_N,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(222),plt.imshow(imgN,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])

plt.subplot(223),plt.imshow(img_P,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(224),plt.imshow(imgP,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])

plt.show()
```

# 15. Buatlah program skeleton pada citra biner, lalu tampilkan citra biner dan rangkanya secara berdampingan.

```python
import cv2 as cv
from matplotlib import pyplot as plt
from skimage import filters
from skimage.morphology import skeletonize

# read image
img = cv.imread('/content/drive/MyDrive/PCVK/Images/lily.jpg',0)

binary = img > filters.threshold_triangle(img)

# true false to one
binary_cp = binary.copy()
binary_cp[binary_cp == True] = 1
binary_cp[binary_cp == False] = 0

# skeletonize image
skeleton = skeletonize(binary_cp)

# print images
plt.figure(figsize=(20,20))

plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])

plt.subplot(122),plt.imshow(skeleton,cmap = 'gray')
plt.title('skeleton Image'), plt.xticks([]), plt.yticks([])

plt.savefig("binary and bin_skeleton.png")
plt.show()
```
